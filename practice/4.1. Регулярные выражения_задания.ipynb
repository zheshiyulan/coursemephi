🔹 **Задание 1. Разбить строку на список чистых слов/словосочетаний без символов**

```python
import re
```

```python
text = "Nice!Good job#Well done- Perfect"
```

```python
split = re.split(r"[\!#-]", text)
split
```

🔹 **Задание 2. Замаскировать только последние четыре цифры номера карты на звездочки**

```python
text = "Номер карты: 1234 5678 9012 3456"
```

```python
sub = re.sub(r"\d{4} \d{4} \d{4} \d{4}", "*", text)
sub
```

🔹 **Задание 3. Вывести только первую фамилию с окончанием "ий"**

```python
text = "В списке фамилий были Иванов, Яновский и Васильев"
```

```python
search = re.search(r"[А-ЯЁ][а-яё]+ий", text)
print(search)
print(search.group(0))
```

🔹 **Задание 4. Вывести весь текст между комментариями, то есть между звездочками**

Примечание: не забыть, что нужно экранировать некоторые симолвы (обратный слэш)

```python
text = "код * исправить вот здесь! добавить проверку * продолжение"
```

```python
search = re.search(r"\s *()*", text)
print(search.group(1))
```

🔹 **Задание 5. Вывести все цены из текста**

Примечание: цена может быть из любого количество цифр и может быть указана с копейками, то есть от 1 до 2 знаков после точки

```python
text = "Количество: 5 штук. Цена: 199.90 рублей. Завтра будет цена: 220 рублей"
```

```python
findall = re.findall(r"\d{3}\.+\d\", text)
findall
```

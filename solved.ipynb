```python
# надо смотреть выбросы и тип данных (категориальные или нет), чтобы понять, чем заполнить пропуски. 
# если еще и выброс в числовом, то лучше использовать медиану. выброса нет - среднее. для категориальных - мода. 
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=df['Ease of Online booking'])
plt.show()
```

```python
# выбросов нет, числовое => mean
mean_value_booking = df['Ease of Online booking'].mean()
df['Ease of Online booking'].fillna(mean_value_booking, inplace=True)
df.isna().sum()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=df['Gender'])
plt.show()
```

```python
# выбросов нет, категориальное => мода
mode_value_gender = df['Gender'].mode()[0]
df['Gender'].fillna(mode_value_gender, inplace=True)
df.isna().sum()
```

```python
df.boxplot(column=['Leg room service']) 
# тут я нашла ящик покрасивее, поэтому они разные до и тут. дезигн
```

```python
# снова нет выбросов, числовое => mean
mean_value_leg = df['Leg room service'].mean()
df['Leg room service'].fillna(mean_value_leg, inplace=True)
df.isna().sum()
```

```python
df.boxplot(column=['Baggage handling'])
```

```python
# опять без выбросов, опять число, опять mean
mean_value_handling = df['Baggage handling'].mean()
df['Baggage handling'].fillna(mean_value_handling, inplace=True)
df.isna().sum()
```

```python
df.boxplot(column=['Arrival Delay in Minutes'])
```

```python
# ура выбросы!!!! но есть нюанс: выбросов много потому что 0 это основной ответ. 
# решение - заменить 0 на nan чтобы не отображалось в боксплоте и потом глянуть реальный выброс
```

```python
df.mask(df==0).boxplot(column=["Arrival Delay in Minutes"])
df.tail()
```

```python
# более менее адекватный выброс, а если он есть и значение числовое, то ищем медиану
```

```python
median_arr_delay = df['Arrival Delay in Minutes'].mean()
df['Arrival Delay in Minutes'].fillna(median_arr_delay, inplace=True)
df.isna().sum()
```

```python
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].replace(np.nan, 0.0)
df.tail()
```

```python
df.describe()
# если mean отличается от 50% (сильно), то есть выбросы.
# в нашем случае есть выбросы по Flight Distance, Departure Delay in Minutes, Arrival Delay in Minutes
```

```python
# очевидно, что от общего среднего балла может зависеть удовлетворенность рейсом, так что введем новую метрику
overall = df[['Inflight wifi service','Departure/Arrival time convenient','Ease of Online booking','Gate location','Food and drink','Online boarding','Seat comfort','Inflight entertainment','On-board service','Leg room service','Baggage handling','Checkin service','Inflight service','Cleanliness']].mean(axis=1)
df.insert(3, 'Overall service', overall)
```

```python
df.head()
# в целом метрика имеет место быть, но даже с высокой оценкой (уже выше 3-3,5) бывают плохие рейсы
# а связано это с задержками. значит, их надо как-то интегрировать в оценку рейса (и ввести штрафные баллы за сильные задержки)
```

```python
overall_final = np.where(
    df['Arrival Delay in Minutes'].isna() | (df['Arrival Delay in Minutes'] == 0),
    overall,
    overall * (1 - np.minimum(df['Arrival Delay in Minutes'] / 500, 0.4)) 
)

df.insert(4, 'Overall final', overall_final)
```

```python
df.tail(20)
```

```python
correlation = df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(20, 16))  
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.tight_layout()
plt.show()
```

```python
# чтобы при обучении модель не сбивалась на странную корреляцию, чуть почищу датасет
df = df.drop('Overall service', axis=1)
df
```

```python
correlation_2 = df.corr(numeric_only=True)
plt.figure(figsize=(20, 16))  
sns.heatmap(correlation_2, annot=True, cmap='coolwarm', fmt='.2f')
plt.tight_layout()
plt.show()
```

```python
# но лично мне больше интересна корелляция между оценкой рейса (уже с учетом задержки!) и satisfied
df['satisfaction_numeric'] = df['satisfaction'].map({'satisfied': 1, 'neutral or dissatisfied': 0})

correlation = df[['Overall final', 'satisfaction_numeric']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.3f', square=True)
plt.title('Overall final и Satisfaction')
plt.show()
# она между 3 и 5, значит все таки есть какая-то связь, гештальт закрыт
```
